Initial Module imports

```python
from core.node.manager import NodeManager
from core.session import SessionManager
available_nodes = NodeManager.available_nodes()
print(available_nodes)
```

Create Node instances of nodes you want

```python
exchange = NodeManager.create_node_instance('exchange', 'KuCoin')
rsi = NodeManager.create_node_instance('dataprocessor', 'RSI')
```

Create a Session Manager

```python
exchange.get_in_io()
```

```python
ssm = SessionManager()
```

```python
project = ssm.start_new_project("SandBox Project")
```

```python
editor = project.worker
```

```python
editor.attach_node(exchange)
editor.attach_node(rsi)
```

```python
dp = exchange.get_dynamic_params()
sp = exchange.get_static_params()
print(f"Dynamic Parameters: {dp}")
print(f"Static Parameters: {sp}")
```

```python
editor.connect_node_io(
    output_node_name=exchange.name,
    output_port='ohlcv',
    input_node_name=rsi.name,
    input_port='ohlcv'
)
```

```python
editor.build_trnx()
trnx_bot = editor.get_trnx()
```

```python
ssm.start_new_trnx_session("MyTrnx")
```

```python
ssm.retrieve_session_by_name('MyTrnx')
```

```python
trnx_session = ssm.retrieve_session_by_name('MyTrnx')
```

```python
trnx_session.load_trnx(trnx_bot)
```
